## 8.3 — Falsifiable Scientific Critic contract

**Decisions:** `evidence_only` (a narrowly authorized, canonical published observation), `hypothesis_only` (a proposed *unrelaxed* registered geometry requiring review), or `abstain` (insufficient/contradictory support).

This critic uses **structured inputs** rather than accepting arbitrary prose with a valid citation appended. Reported FE is checked against **exact catalyst, product, numerical FE and potential**. Product-ratio claims can use the reviewed source paraphrase only; ratio numbers are not independently structured in this initial record. Open-ended LLM text cannot be automatically approved.

This is a software and policy critic, not an independently trained scientific judge.

```python
# PSEUDOCODE
# 1. Define an immutable structured scientific assertion and critic decision.
# 2. Specify the only allowed claim types and narrow approval categories.
# 3. Implement safe normalization of catalyst identity, measured FE and electrode potential.
# 4. Build canonical responses from the reviewed record, not from freeform proposed text.

ClaimKind08 = Literal[
    'reported_fe','reported_ratio','structure_hypothesis','catalyst_superiority',
    'energy_prediction','mechanistic_certainty','durability_claim','evaluation_claim',
    'instructions_as_evidence'
]
Decision08 = Literal['evidence_only','hypothesis_only','abstain']

@dataclass(frozen=True)
class ScientificAssertion08:
    assertion_id: str
    kind: ClaimKind08
    evidence_ids: tuple[str, ...] = ()
    candidate_ids: tuple[str, ...] = ()
    catalyst: str = ''
    product: str = ''
    fe_percent: float | None = None
    potential_v_rhe: float | None = None
    statement: str = ''

@dataclass(frozen=True)
class CriticDecision08:
    assertion_id: str
    decision: Decision08
    reason_codes: tuple[str, ...]
    cited_ids: tuple[str, ...] = ()
    candidate_ids: tuple[str, ...] = ()
    canonical_response: str = ''
    scientific_modeling_approved: bool = False

SUPPORTED_TYPES08 = {'reported_fe','reported_ratio','structure_hypothesis'}
BLOCKED_TYPES08 = {'catalyst_superiority','energy_prediction','mechanistic_certainty',
                   'durability_claim','evaluation_claim','instructions_as_evidence'}

# This baseline requires the original, exact human-verified claim wording.
def text_key08(text):
    return ' '.join(str(text).strip().split())

def potential08(conditions):
    standardized = str(conditions).replace('−','-').replace('–','-')
    hits = re.findall(r'(?<![\w.])([+-]?\d+(?:\.\d+)?)\s*V\s*vs\s*RHE',
                      standardized, flags=re.IGNORECASE)
    return float(hits[0]) if len(hits) == 1 else None

def abstain08(a, code):
    return CriticDecision08(a.assertion_id,'abstain',(code,))

print('SCIENTIFIC CRITIC CONTRACT: DEFINED')
print('Supported narrow claims:',sorted(SUPPORTED_TYPES08))
print('Unvalidated assertion types blocked:',sorted(BLOCKED_TYPES08))
print('Human approvals and untrusted text cannot change execution policy')
```

## 8.4 — Evidence- and geometry-constrained Critic implementation

A valid citation alone **does not** support a claim about a different product, catalyst, potential, or value. The Critic fails closed on multiple citations for one reported FE, missing fields, or arbitrary alternative wording. It **never** promotes geometry hypotheses to performance facts.

```python
# PSEUDOCODE
# 1. Validate requested evidence identifiers and candidate identifiers against authoritative records.
# 2. Block unsupported assertion categories and any noncanonical scientific claim wording.
# 3. For FE, verify the reported product, catalyst, FE and potential against source metadata.
# 4. For ratio, permit only the exact approved paraphrase as a reported observation.
# 5. For geometry hypotheses, require registered, pending-review and unrelaxed candidates.
# 6. Return an immutable outcome with no approval side effects.

def critic08(a: ScientificAssertion08) -> CriticDecision08:
    if not a.assertion_id.strip():
        return abstain08(a,'missing_assertion_id')
    if a.kind in BLOCKED_TYPES08:
        return abstain08(a,'unsupported_scientific_assertion_type')
    if a.kind not in SUPPORTED_TYPES08:
        return abstain08(a,'unknown_assertion_type')
    if len(a.evidence_ids) != len(set(a.evidence_ids)) or len(a.candidate_ids) != len(set(a.candidate_ids)):
        return abstain08(a,'duplicate_reference_id')
    if not set(a.evidence_ids) <= set(corpus_by_id08):
        return abstain08(a,'unapproved_or_fabricated_citation')
    if not set(a.candidate_ids) <= registered_ids08:
        return abstain08(a,'unknown_geometry')

    if a.kind == 'structure_hypothesis':
        if a.evidence_ids or not (1 <= len(a.candidate_ids) <= MAX_CRITIC_CANDIDATES08):
            return abstain08(a,'invalid_hypothesis_input')
        if any(term in a.statement.lower() for term in (
            'best','better','superior','stable','selective','predicted','guaranteed','higher fe'
        )):
            return abstain08(a,'performance_language_in_geometry_hypothesis')
        by_candidate = registry08.set_index('candidate_id')
        for cid in a.candidate_ids:
            r=by_candidate.loc[cid]
            if (r['geometry_status']!='unrelaxed' or
                r['structure_review_status']!='pending_visual_review' or
                r['approved_for_modeling'].strip().lower()!='false'):
                return abstain08(a,'unreviewed_geometry_not_eligible_for_approval')
        return CriticDecision08(
            a.assertion_id,'hypothesis_only',
            ('unrelaxed_geometry','human_visual_review_and_reference_calculation_required'),
            candidate_ids=a.candidate_ids,
            canonical_response='Registered Cu–Ag geometry hypothesis; no stability, energy, or selectivity claim.'
        )

    if a.candidate_ids or len(a.evidence_ids)!=1:
        return abstain08(a,'exactly_one_evidence_id_required')
    source=corpus_by_id08[a.evidence_ids[0]]
    if a.statement and text_key08(a.statement) != text_key08(source['claim_paraphrase']):
        return abstain08(a,'noncanonical_scientific_text_requires_review')
    if source['review_status'] != 'human_verified':
        return abstain08(a,'claim_not_human_verified')
    if source['paper_id'] not in paper_doi08 or source['doi'].lower().strip()!=paper_doi08[source['paper_id']]:
        return abstain08(a,'invalid_publication_link')
    if a.kind == 'reported_fe':
        if not source['fe_product'].strip() or not source['faradaic_efficiency_percent'].strip():
            return abstain08(a,'not_a_structured_fe_measurement')
        if a.catalyst.strip() != source['catalyst_system'].strip():
            return abstain08(a,'catalyst_identity_mismatch')
        if a.product.strip().upper() != source['fe_product'].strip().upper():
            return abstain08(a,'product_mismatch')
        actual_v=potential08(source['conditions'])
        if actual_v is None or a.fe_percent is None or a.potential_v_rhe is None:
            return abstain08(a,'required_numerical_metadata_missing')
        if not (math.isfinite(a.fe_percent) and math.isclose(a.fe_percent,float(source['faradaic_efficiency_percent']),abs_tol=1e-9)):
            return abstain08(a,'faradaic_efficiency_mismatch')
        if not (math.isfinite(a.potential_v_rhe) and math.isclose(a.potential_v_rhe,actual_v,abs_tol=1e-9)):
            return abstain08(a,'potential_mismatch')
        response=(f"{source['claim_paraphrase']} [{source['evidence_id']}] "
                  f"DOI: {source['doi']}; source: {source['source_extent']}, "
                  f"{source['source_location']}; conditions: {source['conditions']}.")
        return CriticDecision08(a.assertion_id,'evidence_only',('exact_source_match','reported_fe_not_prediction'),
                                 cited_ids=a.evidence_ids,canonical_response=response)

    # The ratio source is human-reviewed text, not a structured FE row.
    if a.kind == 'reported_ratio':
        if source['fe_product'].strip() or source['faradaic_efficiency_percent'].strip():
            return abstain08(a,'ratio_not_a_faradic_efficiency')
        if 'ratio' not in source['target_or_observation'].lower():
            return abstain08(a,'ratio_observation_mismatch')
        if (a.fe_percent is not None or a.potential_v_rhe is not None or a.product.strip()):
            return abstain08(a,'unstructured_ratio_cannot_gain_fe_or_potential')
        if a.catalyst and a.catalyst.strip()!=source['catalyst_system'].strip():
            return abstain08(a,'catalyst_identity_mismatch')
        response=(f"{source['claim_paraphrase']} [{source['evidence_id']}] "
                  f"DOI: {source['doi']}; source: {source['source_extent']}, "
                  f"{source['source_location']}. No structured FE or potential inferred.")
        return CriticDecision08(a.assertion_id,'evidence_only',('exact_reviewed_paraphrase','ratio_not_fe'),
                                 cited_ids=a.evidence_ids,canonical_response=response)
    return abstain08(a,'fail_closed')

print('FALSIFICATION GATE: DEFINED (read-only, deterministic)')
```

## 8.7 — Replay and no-mutation controls

A deterministic policy gate should reproduce its answer on repeated calls and never modify human review records. These checks operate **on immutable objects and in-memory copies**. They do not establish correctness for unseen publications or arbitrary natural-language generation.

## 8.9 — Active-learning precursor: evidence and calculation triage

**This is NOT Bayesian active learning.** We have neither a validated predictive model nor a calibrated uncertainty distribution. Therefore, an expected-information-gain or acquisition-function claim would be fabricated.

Instead, we create an **auditable, human-directed *research-action triage***. Ordinal 0–3 scores for gap relevance and practical feasibility and a 0–3 prerequisite penalty are **developer-authored priorities**, not energies, FE, probabilities, ML uncertainties, or measured information gain.

**Rules:**
- A high-priority row is a suggestion for a **research task**, never a claim about the best catalyst.
- Reference DFT calculations require separate human review and computational setup.
- A publication-held-out evaluation needs more than one evidence-bearing publication.
- No action is executed automatically. No new evidence is approved automatically.

```python
# PSEUDOCODE
# 1. List research tasks addressing real gaps found in Notebooks 02–07.
# 2. Assign transparent, manually chosen ordinal triage scores (not learned uncertainty).
# 3. Mark prerequisite-blocked actions without automatically approving or running them.
# 4. Rank research *actions*, not catalysts, with deterministic tie-breaking.
# 5. Display the rationale and limits of the heuristic.

research_actions08=[
    dict(action_id='R08-01',action='Human-review independent Cu–Ag CO2RR experimental publications',
         kind='source_review',gap_relevance=3,feasibility=3,prerequisite_penalty=0,
         prerequisites='source access; independent human scientific review',
         status='propose_for_manual_planning'),
    dict(action_id='R08-02',action='Complete human visual review of eight exported slab models',
         kind='geometry_review',gap_relevance=3,feasibility=3,prerequisite_penalty=0,
         prerequisites='human confirmation of layer positions and periodic motifs',
         status='propose_for_manual_planning'),
    dict(action_id='R08-03',action='Structure experimental condition metadata for P005 observations',
         kind='condition_curation',gap_relevance=3,feasibility=2,prerequisite_penalty=0,
         prerequisites='source verification; no inferred missing operating conditions',
         status='propose_for_manual_planning'),
    dict(action_id='R08-04',action='Acquire independent reference DFT energies/forces for matched Cu(100) Ag pairs',
         kind='reference_calculation_request',gap_relevance=3,feasibility=1,prerequisite_penalty=2,
         prerequisites='human geometry approval; defined method; compute budget; convergence tests',
         status='blocked_pending_prerequisites'),
    dict(action_id='R08-05',action='Design DOI-held-out retrieval and critic evaluation using multiple publications',
         kind='benchmark_design',gap_relevance=3,feasibility=1,prerequisite_penalty=2,
         prerequisites='multiple independently reviewed publications; frozen evaluation questions',
         status='blocked_pending_prerequisites'),
    dict(action_id='R08-06',action='Review pretrained MLIP Cu/Ag surface domain applicability',
         kind='model_applicability_review',gap_relevance=2,feasibility=2,prerequisite_penalty=1,
         prerequisites='training domain documentation; relevant DFT references',
         status='blocked_pending_prerequisites'),
]
for r in research_actions08:
    assert all(isinstance(r[f],int) and 0<=r[f]<=3
               for f in ('gap_relevance','feasibility','prerequisite_penalty'))
    r['triage_score_not_scientific']=2*r['gap_relevance']+r['feasibility']-2*r['prerequisite_penalty']
action_queue08=pd.DataFrame(research_actions08).sort_values(
    ['triage_score_not_scientific','action_id'],ascending=[False,True]
).reset_index(drop=True)
assert action_queue08['action_id'].is_unique and len(action_queue08)==6
assert not action_queue08['kind'].eq('catalyst_ranking').any()
display(action_queue08[['action_id','action','kind','triage_score_not_scientific','status','prerequisites']])
print('RESEARCH TRIAGE: 6 developer-authored tasks; 0 tasks executed')
print('Scores are heuristic planning priorities, NOT uncertainty or model predictions')
```
